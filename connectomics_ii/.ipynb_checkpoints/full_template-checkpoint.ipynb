## Single brain areas are very nice, but let's do it for all the brain areas.

#### look at the code above and adapt it to work in a loop for all the neuropils not just one
* instantiate variables that catch the average connectivity and reciprocity
* create a loop that iterates over all brain areas
* calculate the metrics store them in the variables
* plot the variables for all the brain areas a bar plot
  
  

```python
def conn_prob(G):
    N = G.number_of_nodes()
    E = G.number_of_edges()
    return E/(N*(N-1))

#delete after here ///////////////////////////////////////////////
all_areas = np.unique(raw_data["neuropil"])
avg_con = []

recip = []
sizes = []
for i in all_areas:
    
    idat = raw_data[raw_data["neuropil"] == i]
    
    Gi = nx.from_pandas_edgelist(
        idat,
        source='pre_root_id',
        target='post_root_id',
        edge_attr='weight',
        create_using=nx.DiGraph()
    )

    sizes.append(len(list(Gi.nodes)))
    avg_con.append(conn_prob(Gi))
    recip.append(nx.reciprocity(Gi))

#append the whole system at the end. 
sizes.append(len(list(G_dir_dros.nodes)))
avg_con.append(conn_prob(G_dir_dros))
recip.append(nx.reciprocity(G_dir_dros))
all_areas = np.hstack((all_areas,np.array(['all'])))


# Do the visualiztion 
print('######################################')

fig, ax = plt.subplots(figsize=(12,2))
ax.bar(np.arange(len(avg_con)),avg_con)
ax.set_ylabel('avg. conn. prop.')
ax.set_xticks(np.arange(len(avg_con)))
ax.set_xticklabels(all_areas,rotation=90)
ax.axhline(y=avg_con[-1],color='red')
ax.set_yscale('log', base=10)
plt.show()

fig, ax = plt.subplots(figsize=(12,2))
ax.bar(np.arange(len(sizes)),sizes)
ax.set_ylabel('num neurons')
ax.set_xticks(np.arange(len(sizes)))
ax.set_xticklabels(all_areas,rotation=90)
plt.show()

fig, ax = plt.subplots(figsize=(12,2))
ax.bar(np.arange(len(recip)),recip)
ax.set_ylabel('reciprocity')
ax.set_xticks(np.arange(len(avg_con)))
ax.set_xticklabels(all_areas,rotation=90)
plt.show()




```

# Let's build a histogram function that shows all the counts for the individual classes occuring in the SCCs labels

### implement a function that: 
* takes both SCCs as input
* indexes all the unique classes within the labels.
* counts the number of occurances for each class from the individual scc
* plot the histogramms side-by-side  

```python
def class_histogram(arr1, arr2, title1="component 1", title2="component 2"):
  
    if not isinstance(arr1,np.ndarray):
        arr1 = np.array(arr1)
    if not isinstance(arr2,np.ndarray):
        arr2 = np.array(arr2)

     # list all unique classes
    classes = sorted(set(arr1) | set(arr2))
    # counts
    counts1 = [np.sum(arr1 == c) for c in classes]
    counts2 = [np.sum(arr2 == c) for c in classes]
    fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)
    x = np.arange(len(classes))  # explicit bar positions

    # Histogram 1
    axes[0].bar(x, counts1)
    axes[0].set_title(title1)
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(classes, rotation=90)
    axes[0].set_ylabel("Count")

    # Histogram 2
    axes[1].bar(x, counts2, color="orange")
    axes[1].set_title(title2)
    axes[1].set_xticks(x)
    axes[1].set_xticklabels(classes, rotation=90)

    plt.tight_layout()
    plt.show()


#print(endings1)
#print(endings2)
class_histogram(endings1, endings2, title1="SCC 1", title2="SCC 2")
```

```python
#this is how we retrieve elements from a dataframe: 
print('Neuron class of AWAR is: ' + classdat.loc[classdat["Neuron"] == "AWAR",'class'])

##delete from here 

def check_class(input_list,df):
    clss = []
    for i in input_list:
        subset = df.loc[df["Neuron"] == i.upper(), "class"]
        #subset = df.query("Neuron.str.upper() == @i.upper()")["class"]
        #subset = df.loc[df["Neuron"].str.casefold() == i.casefold(), "class"]
        if len(subset) > 0:
            clss.append(subset.iloc[0])
        else:
            clss.append("NA")
    return clss

classes1 = check_class(c1,classdat)
classes2 = check_class(c2,classdat)


class_histogram(classes1, classes2, title1="SCC 1", title2="SCC 2")
a1 = np.array(list(c1))
a2 = np.array(list(c2))

print('########################################')
print("sensory neurons in SCC1:")
idx = np.array(classes1) == "sensory neuron"
print(a1[idx])
      
print('########################################')
print("sensory neurons in SCC2:")

idx = np.array(classes2) == "sensory neuron"
print(a2[idx])






print('########################################')
print("Motor neurons in SCC1:")
idx = np.array(classes1) == "motor neuron"
print(a1[idx])
      
print('########################################')
print("Motor neurons in SCC2:")
idx = np.array(classes2) == "motor neuron"
print(a2[idx])

```
